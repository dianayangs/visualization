```python
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0, 100,50)
y2 = np.random.randint(0,100,50)
```

```python
fig,ax = plt.subplots(figsize=(5,3)) #creating the object
ax.plot(x,y1)
ax.plot(x,y2)
fig.show() #can also use plt.show() 
```

```python
#adding labels and legend
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y1, label="Person 1") #adding legend
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='lower right',
        frameon = True, #add frame around the legend
        fontsize = 12, #change font size
        ncol = 2, #specify number of columns to make it horizontal instead of vertical stack
        shadow = True)
fig.show()
```

```python
#same as above but legend is outside
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y1, label="Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='upper left', #attaching upper left corner of legend to upper right corner of plot when used in conjuction with bbox_to_anchor() fxn
    bbox_to_anchor =(1, 1))
#trying the text on this plot as well
ax.text(10, 95, "This value is important!",
        ha='center', ##alignment of text
        color = 'red', ##modify font colour
        size = 20) ##modify font size #adding annotations with modifications
fig.show()
```

```python
# labelling scatter plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.text(10, 95, "This value is important!",
ha='center', ##alignment of text
color = 'red', ##modify font colour
size = 20) ##modify font size
#fig.show()
```

```python
#using transforms to position text/data
fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])
ax.text(1, 5, ". Data:(1, 5)", transform=ax.transData )
ax.text(0.5, 0.1, ". Axes:(0.5, 0.1)", transform=ax.transAxes )
ax.text(0.2, 0.2, ". Figure:(0.2, 0.2)", transform=fig.transFigure )
```

```python
#now we annotate with arrows and text with annotate()
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!', xy=(10, 95), xytext=(20, 94),
    arrowprops=dict(facecolor='black'))
fig.show()
```

```python
#we can also modify arrows using arrowprop = 
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!',
    xy=(10, 95), xytext=(20, 94),
    arrowprops = dict(arrowstyle = 'wedge',
        color = 'hotpink'))
fig.show()
```

```python
#remove tick marks and labels using locator and formatter
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.yaxis.set_major_locator(plt.NullLocator())
ax.xaxis.set_major_formatter(plt.NullFormatter())

#this graph has no axis labels and y axis has no ticks. cleaner look
```

```python
#or define a max number of ticks
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MaxNLocator(3))
```

```python
#setting intervals using MultipleLocator
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MultipleLocator(5)) #ticks every 5 on the x axis
```

```python
#rotate axis labels
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
plt.xticks(rotation=45, ha='right', fontsize=20) #text is now at an angle, cool! also size is now huge
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75,50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
fig, ax = plt.subplots(figsize=(5, 3)) #one plot
fig, (ax1, ax2) = plt.subplots(ncols=2, #2 plots
nrows=1,
figsize=(7, 3))
ax1.scatter(x1,y1)
ax2.bar(x2,y2)
fig.show()
```

```python
#modifying subplot
plt.style.use('bmh') 
fig, (ax1, ax2) = plt.subplots(ncols=2, #2 plots
nrows=1,
figsize=(7, 3))
ax1.scatter(x1,y1)
ax2.bar(x2,y2)
fig.show()
```

```python
#arranging subplots in a figure:
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
['ax2', 'ax3']], #order: left to right, top to bottom. noting ax3 twice will ensure that the right most graph is just one and fills the entire matrix space
figsize=(7, 4))
```

```python
#adding data to new subplot
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
['ax2', 'ax3']],
figsize=(7, 4),
layout="constrained") #adding constraint will make all labels visible
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
#add large labels
someaxes["ax1"].set_xlabel('A Big Label',fontsize=18) #too big doesn't show when you don't use layout='constrained'
someaxes["ax2"].set_xlabel('Another Label',fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious',fontsize=18 )
plt.show()
```

```python
#multiple viz on one axes
#sample data
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])
# define our figure and axes
fig, ax = plt.subplots(figsize=(7, 3))
# now call both bar and plot elements to the same axes (ax)
ax.bar(x, y1,
color = "indigo")
ax.plot(x, y2,
color = "red")
```

```python
#adding error info on the plot above:
#calculate error (SD)
y2_sd = np.std(y2)
#plot
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")
#add errorbars
ax.errorbar(x, #our x values
y2, #our y values
yerr = y2_sd,
fmt = "none",
ecolor='indigo', #colour of error bar
elinewidth=4, #thickness of error bar
capsize=6, #error bar cap width
capthick=4,#error bar cap thickness
errorevery=2) #intervals for error bars
```

```python
#first load libraries
from PIL import Image
import requests
from io import BytesIO
```

```python
url = "https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png"

headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "image/*,*/*;q=0.8",
    "Referer": "https://en.wikipedia.org/"
}

response=requests.get(url,headers=headers)
image_file=BytesIO(response.content)
image=Image.open(image_file)
image
```

```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
tips = sns.load_dataset("tips")
print(tips)

#simple line plot of tip amount vs. total bill
sns.lineplot(data=tips, # choose our dataset
x='total_bill', # define our x variable
y='tip') #define our y variable
```

```python
sns.set_style('whitegrid')
fig = plt.subplots(figsize=(10, 3)) #doing this stretches out the graph but the graph will still populate without
tipgraph=sns.lineplot(data=tips,x='total_bill',y='tip',
                      color='hotpink',
              linestyle='--',
              linewidth=3,
              marker='o',
              markerfacecolor='indigo')
tipgraph.set(title='Tips vs. Total Bill',
              xlabel='Total Bill ($)',
              ylabel='Tip Amount ($)'
              )

```

```python
tipgraph=sns.scatterplot(data=tips,x='total_bill',y='tip',
    style='time', hue='day',
    palette=['purple','hotpink','deepskyblue','yellowgreen'])
tipgraph.set(title='Tips vs. Total Bill',
    xlabel='Total Bill ($)',
    ylabel='Tip Amount ($)')
```

```python
sns.pairplot(
    data=tips,
    hue='day'
)
```

```python
#relpot: allows exploration of specific levels within variables
daysplot=sns.relplot(data=tips,
x="total_bill", #labels x title
y="tip", #y title
hue="sex", #colours each bill by gender of payer
col="day", #each plot is sorted by date
kind="scatter", #scatterplot
col_wrap=2) #2x2 figure
```
