```python
#importing libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
import seaborn as sns
```

```python
#loading data
shelterdata = pd.read_csv(r'/Users/dianayang/Desktop/visualization/02_activities/assignments/participation/viz1_dataappendix.csv')
shelterdata.head
```

```python
#creating variables for the graph
sector, rate = shelterdata.iloc[:,0], shelterdata.iloc[:,1]
```

```python
#loading colour palette
plt.style.use('seaborn-v0_8-colorblind')
#creating font dictionary
font1 = {'family':'serif','color':'indigo'}
#making each bar a different colour
colours= plt.cm.tab10(range(len(sector)))
#plotting
sns.barplot(x=sector, y=rate, palette='viridis', alpha=0.5) #bar plot
#overlay each point
sns.stripplot(x=sector, y=rate, palette='viridis',alpha=0.5)
plt.xlabel('Sector', fontsize=12, fontdict=font1) #labels
plt.ylabel('Occupancy rate (%)', fontsize=12,fontdict=font1)
plt.title('Bed occupancy rate by sector') #title
```
